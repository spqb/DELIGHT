```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
fname_unaligned = "../experiments/datasets/RR/PF00072_unaligned.csv"
fname_aligned = "../experiments/datasets/RR/RR.csv"

df_aligned = pd.read_csv(fname_aligned)
df_unaligned = pd.read_csv(fname_unaligned)

print("Aligned data shape:", df_aligned.shape)
print("Unaligned data shape:", df_unaligned.shape)
```

```python
df_unaligned.head()
```

```python
# extract sequence identifiers from the 'header' column in the unaligned dataframe
def get_id(header):
    return header.split("|")[1]

df_unaligned.rename(columns={"header": "full_header"}, inplace=True)
df_unaligned["header"] = df_unaligned["full_header"].apply(get_id)
```

```python
# join the two dataframes on the 'header' column. Change the 'sequence' column of the aligned dataframe to 'sequence_align' to avoid confusion. Discard rows with no match in the unaligned dataframe.
df_merged = pd.merge(df_aligned.rename(columns={"sequence": "sequence_align"}), df_unaligned, on="header", how="inner")
print("Merged data shape:", df_merged.shape)
```

```python
df_merge = df_aligned.rename(columns={"sequence": "sequence_align"}).merge(df_unaligned, on=["header", "label"], how="inner").filter(items=['header', 'sequence_align', 'sequence', 'label'])
print("Merged data shape:", df_merge.shape)
df_merge.head()
```

```python
# shuffle and save
df_merge = df_merge.sample(frac=1, random_state=42).reset_index(drop=True)
df_merge.to_csv("../experiments/datasets/RR/RR_merged.csv", index=False)
```
